```python
#Drop customer ID. Key is not needed for modeling
df = df.drop("customerID", axis=1)
```

```python
#Target variable Exploration
def plot_count_with_labels(df, col, title=""):
    ax = sns.countplot(x=col, data=df)
    ax.yaxis.set_major_formatter(mtick.StrMethodFormatter("{x:,.0f}"))
    
    total = len(df)
    for p in ax.patches:
        height = p.get_height()
        ax.text(
            p.get_x() + p.get_width() / 2.,
            height,
            f'{height/total:.1%}', 
            ha="center", va="bottom"
        )
    
    ax.set_title(title if title else f"{col} Distribution")
    plt.xticks(rotation=45)
    plt.show()

#Example usage:
plot_count_with_labels(df, "Churn", "Churn Distribution")
plot_count_with_labels(df, "Contract", "Contract Distribution")
```

Roughly 3/4 of customers have not churned at 73.5%, and 26% have. The dataset is imbalanced. 
Most customer's are on month-to-month contracts, which typically will have a higher churn rate. Majority of the population is at a higher risk of churning.  

```python
def stacked_percent_bar(df, x_col="Contract", hue_col="Churn", title=None):
    #1) % table by row (each contract sums to 1.0)
    pct = (pd.crosstab(df[x_col], df[hue_col], normalize="index")
             .reindex(sorted(df[x_col].unique()), axis=0))  # keep logical order

    #2) Plot stacked %
    ax = pct.plot(kind="bar", stacked=True, figsize=(8,5), legend=True)
    ax.set_ylim(0, 1)
    ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
    ax.set_xlabel(x_col)
    ax.set_ylabel("Percent of customers")
    ax.set_title(title or f"{x_col} vs {hue_col} (Percent stacked)")

    #3) Add % labels inside bars
    for container in ax.containers:
        for bar in container:
            height = bar.get_height()
            if height > 0.03:  # hide tiny slivers
                ax.text(
                    bar.get_x() + bar.get_width()/2,
                    bar.get_y() + height/2,
                    f"{height:.1%}",
                    ha="center", va="center", fontsize=10, color="white"
                )

    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

    #4) Chi-square test (are Contract and Churn related?)
    ct = pd.crosstab(df[x_col], df[hue_col])
    chi2, p, dof, _ = chi2_contingency(ct)
    print(f"Chi-square p-value ( {x_col} vs {hue_col} ): {p:.3e}")

stacked_percent_bar(df, x_col="Contract", hue_col="Churn",
                    title="Churn share by Contract type")
```

Customers on a month-to-month contract have the highest churn rate. 43% of these customers leave. 
Customers on a One year contract are more stable, with only a 11% churn rate.
The Two year contract is the most stable, with a 97% retention rate. 

The Chi-Square value of 5.863e-258 is very small, indicating that contract type and churn rate are highly correlated. 

```python
def stacked_percent_bar(df, x_col, hue_col="Churn", title=None):
    #Percent table
    pct = (pd.crosstab(df[x_col], df[hue_col], normalize="index")
             .reindex(df[x_col].unique(), axis=0))  #preserve order
    
    #Plot stacked bar %
    ax = pct.plot(kind="bar", stacked=True, figsize=(8,5), colormap="Paired")
    ax.set_ylim(0, 1)
    ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
    ax.set_xlabel(x_col)
    ax.set_ylabel("Percent of customers")
    ax.set_title(title or f"{x_col} vs {hue_col} (Percent stacked)")
    
    #Add percentage labels inside bars
    for container in ax.containers:
        for bar in container:
            height = bar.get_height()
            if height > 0.03:  #avoid labeling tiny segments
                ax.text(
                    bar.get_x() + bar.get_width()/2,
                    bar.get_y() + height/2,
                    f"{height:.1%}",
                    ha="center", va="center", fontsize=10, color="white"
                )
    
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

    #Chi-square test
    ct = pd.crosstab(df[x_col], df[hue_col])
    chi2, p, _, _ = chi2_contingency(ct)
    print(f"Chi-square p-value ({x_col} vs {hue_col}): {p:.3e}")
```

```python
stacked_percent_bar(df, "InternetService", "Churn", "Churn by Internet Service")
stacked_percent_bar(df, "PaymentMethod", "Churn", "Churn by Payment Method")
stacked_percent_bar(df, "OnlineSecurity", "Churn", "Churn by Online Security")
stacked_percent_bar(df, "TechSupport", "Churn", "Churn by Tech Support")
stacked_percent_bar(df, "PaperlessBilling", "Churn", "Churn by Paperless Billing")
```

Customers with Fiber-Optic internet service have the highest churn rate of 42%.This is substantially more than those with DSL at 19%, or those without internet at 7%.

Customers without online secuirty reflected a higher churn rate than those without, accounting for roughly 42%. The churn rate is much lower, 14% for customers who are enrolled in online secuirty. This trend follows for customers with added technical support services. 

Customers with paperless billing appear to churn more frequently, roughly 33% opposed to those signed to receive paper bills. This may be a reflection of underlying customer types, as month-month customers do not receive paper statements. 

Summary:
Add-on products like online secuirty and tech support strongly reduce churn when paired with Fiber-Optic internet. While churn does appear to be correlated to paperless billing, we need to research trends within underlying customer type to determine if this assumption is accruate. 

```python
#Researching relationship between bill type and contract type
#Create contingency table (cross-tabulation) between paperless billing and contract type

pd.crosstab(df["PaperlessBilling"], df["Contract"], normalize="index") * 100
#Converts raw counts into percentages of each PaperlessBilling group.

#The majority of paperless billing customer are month-to-month as expected.
#This shows that paperless billing is not the cause of churn, but rather contract type is.
```

```python
#Visualize relationship between paperless billing across contract types

sns.countplot(x="Contract", hue="PaperlessBilling", data=df)
plt.title("Paperless Billing Distribution across Contract Types")
plt.show()

#Paperless billing is strongly correlated to month to month contracts. 
```

```python
#Research relationship between bill type and payment type
#Create contingency table (cross-tabulation) between paperless billing and payment type

pd.crosstab(df["PaperlessBilling"], df["PaymentMethod"], normalize="index") * 100

#Amoung paperless billing, 42% of customers pay via Electronic check. 
#Amoung non-paperless billing, only 21% pay electronically, while 33% still mail physical checks.
#Automatic methods (bank transfer/credit card) are roughly similar between both groups (21–22 %), so they don’t drive the difference.
```

The 'paperless customers churn more often' effect is not casual. It’s a correlation explained by payment and contract type behavior. people who pay electronically and manage short-term contracts are inherently more flexible (and more likely to switch providers).
