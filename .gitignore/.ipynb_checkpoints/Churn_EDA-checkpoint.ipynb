Customer Churn -  Exploratory Data Analysis

Dataset Used: https://www.kaggle.com/datasets/blastchar/telco-customer-churn?resource=download

Business Goal: The objective of this project is to utilize the unfamiliar dataset to determine underlying relationships between variables where customer accounts churned. This infromation will be used in predictive machine learning algorithms to model the probability that a customer will cancel based on their displaying characteristics. This infromation is useful to leadership as it will allow for us to identify at risk customers early on and proactivley attempt to attain them.
